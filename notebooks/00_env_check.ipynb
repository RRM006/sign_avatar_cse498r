# 00 — Environment check (Day 1, Step 1)

Checks that this machine can run the pipeline and writes `work/reports/env_report.json`.

- **Run it with** the project's `.venv` kernel (VS Code: *Select Kernel* → `.venv`).
- **What it checks:** Python and package versions; MediaPipe HolisticLandmarker (model file, result fields, IMAGE and VIDEO mode); OpenCV rotation on two sample clips; PyTorch devices; write permission; a small CPU benchmark.
- **What it does not do:** install, download, or change any dataset file (the two clips are only read, and their SHA-256 is compared before and after).
- **Where:** set `BDSL_ENV` (`local` / `kaggle` / `colab`) or let it auto-detect; `BDSL_WORK_DIR` moves the output folder.

```python
from bdsl.config import load_config

cfg = load_config()  # env layer auto-detected; no data layer needed for this check
print("env:", cfg["_meta"]["env"], "| config hash:", cfg["_meta"]["config_hash"][:12])
print("work dir:", cfg["paths"]["work_dir"])
```

```python
from bdsl.env import run_env_check

report = run_env_check(cfg)
print("overall:", report["overall"], "| written to:", report["written_to"])
for name, status in report["summary"].items():
    print(f"  {name:<20} {status}")
```

```python
# Details of anything that did not pass (empty output = all passed).
import json

for name, check in report["checks"].items():
    if check.get("status") != "pass":
        print(name, json.dumps(check, indent=2, ensure_ascii=False)[:2000])
```

**Reading the result.** `overall: pass` means every check passed on this machine. Any `fail` keeps its error in the report; nothing is skipped silently. The benchmark is a single-process smoke measurement, not an estimate for the full dataset.

# 06 — Avatar demo: one sentence → Model B and B1 side by side (Day 5)

Runs exactly the same code as `uv run python -m bdsl.cli demo --sentence "…"` (see `CLAUDE.md` → COMMANDS).

- **Text → gloss:** the G0 rule drafter (Model A only ran in TEST mode, ADR-0025).
- **Two motions, both labelled (G11, owner 2026-10-22):** Model B (pilot, 40 training clips) and B1 retrieval (the real recording of the most similar *training* sentence).
- **Outputs** in `work/demo/demo_<time>/`: `motion_modelB.json`, `motion_B1.json`, `skeleton_side_by_side.mp4` (always), `demo_log.json` (versions, hashes, round-trip errors), `LIMITATIONS.md`.
- **Avatar viewer:** `uv run python -m http.server 8765 --bind 127.0.0.1` from the project root, then open `http://127.0.0.1:8765/web/avatar_viewer/?demo=/work/demo/<demo folder>/`. It needs the CC0 VRoid model in `work/models/vrm/Base_Female.vrm` (gitignored) and internet access for three.js / three-vrm (jsDelivr, pinned).
- **Not checked by a BdSL-competent person.** Read `LIMITATIONS.md` before showing it anywhere.


```python
from bdsl.cli import main
from bdsl.config import load_config

cfg = load_config(data="pilot50")
print(cfg["avatar"])
```

```python
# Pilot test sentence 3246 (owner's choice); any Bangla sentence works, unknown words become <unk>.
from bdsl.text.annotations import read_csv

ann = read_csv(cfg["paths"]["annotations_dir"] + "/" + cfg["annotations"]["pilot_file"])
sentence = ann.loc[ann["index"] == "3246", "sentence_raw"].iloc[0]
print("exit code:", main(["demo", "--sentence", sentence]))
```

```python
# The newest demo: log summary and the skeleton video.
import json
from pathlib import Path

from IPython.display import Video

demo = max(Path(cfg["paths"]["demo_dir"]).glob("demo_*"), key=lambda p: p.stat().st_mtime)
log = json.loads((demo / "demo_log.json").read_text(encoding="utf-8"))
print(json.dumps({k: log[k] for k in ("sentence", "g0_gloss", "b1", "roundtrip_pilot50_deg")}, ensure_ascii=False, indent=1))
Video(str(demo / "skeleton_side_by_side.mp4"), embed=True)
```
